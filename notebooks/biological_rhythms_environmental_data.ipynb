## 3. Biological Periodicity Analysis

### 3.1 Arabidopsis thaliana (Model Plant)

```python
# Analyze Arabidopsis circadian rhythms
arabidopsis = ArabidopsisAnalyzer()
arab_results = arabidopsis.analyze_all_periods()

print("Arabidopsis Periodicity Analysis")
print("="*50)
print(f"Species: {arab_results['species']}")
print(f"Fundamental frequency (f₀): {arab_results['f0_hz']} Hz\n")

for rhythm_name, analysis in arab_results['periods'].items():
    print(f"\n{rhythm_name.upper()}:")
    print(f"  Period: {analysis['biological_period_hours']:.2f} hours")
    print(f"  Frequency: {analysis['biological_freq_hz']:.6e} Hz")
    print(f"  Harmonic ratio: {analysis['harmonic_ratio']:.2f}")
    print(f"  Nearest harmonic: {analysis['nearest_harmonic']}")
    print(f"  Deviation: {analysis['harmonic_deviation_percent']:.4f}%")
    print(f"  Is harmonic: {'✓ YES' if analysis['is_harmonic'] else '✗ NO'}")
```

### 3.2 Trichogramma (Parasitoid Wasp)

```python
# Analyze Trichogramma developmental cycles
trichogramma = TrichogrammaAnalyzer()

# Analyze at different temperatures
temperatures = [20, 25, 30]  # Celsius

print("Trichogramma Developmental Analysis")
print("="*50)

for temp in temperatures:
    print(f"\n--- Temperature: {temp}°C ---")
    trich_results = trichogramma.analyze_developmental_stages(temperature_celsius=temp)
    
    for stage_name, analysis in trich_results['stages'].items():
        print(f"\n{stage_name}:")
        print(f"  Period: {analysis['biological_period_hours']:.2f} hours")
        print(f"  Nearest harmonic: {analysis['nearest_harmonic']}")
        print(f"  Deviation: {analysis['harmonic_deviation_percent']:.4f}%")
```

### 3.3 Cross-Species Comparison

```python
# Compare periodicities across species
comparison = compare_species_periodicities()

print("Cross-Species Periodicity Comparison")
print("="*50)
print(f"Fundamental frequency (f₀): {comparison['f0_hz']} Hz\n")

# Count harmonics per species
for species_name, species_data in comparison['species_comparison'].items():
    harmonic_count = sum(
        1 for period_data in species_data['periods'].values()
        if period_data['is_harmonic']
    )
    total_periods = len(species_data['periods'])
    print(f"{species_name.upper()}: {harmonic_count}/{total_periods} periods are harmonic")

print(f"\nTotal harmonic resonances found: {len(comparison['harmonic_resonances'])}")
```

## 4. Environmental Data Integration

### 4.1 NASA POWER Data (Solar/Meteorological)

```python
# Initialize NASA POWER client
nasa_client = NASAPowerAPIClient()

# Example: Fetch data for a research location
# Using Salk Institute coordinates (Arabidopsis research center)
latitude = 32.8875
longitude = -117.2426
location_name = "Salk Institute, La Jolla, CA"

# Fetch one month of data
start_date = '20240101'
end_date = '20240131'

print(f"Fetching NASA POWER data for {location_name}")
print(f"Period: {start_date} to {end_date}")
print("This may take a moment...\n")

nasa_data = nasa_client.get_agricultural_data(
    latitude=latitude,
    longitude=longitude,
    start_date=start_date,
    end_date=end_date
)

if nasa_data and 'parameters' in nasa_data:
    print("✓ Successfully fetched NASA POWER data")
    print(f"Parameters available: {list(nasa_data['parameters'].keys())}")
else:
    print("⚠ Failed to fetch data. Using simulated data for demonstration.")
    # Fallback to simulated data
    dates = pd.date_range(start='2024-01-01', periods=31, freq='D')
    nasa_data = {
        'parameters': {
            'T2M': {str(d.strftime('%Y%m%d')): 15 + 5*np.sin(2*np.pi*i/31) for i, d in enumerate(dates)},
            'ALLSKY_SFC_SW_DWN': {str(d.strftime('%Y%m%d')): 5 + 3*np.sin(2*np.pi*i/31) for i, d in enumerate(dates)}
        }
    }
```

```python
# Visualize environmental data
if 'parameters' in nasa_data:
    fig, axes = plt.subplots(2, 1, figsize=(12, 8))
    
    # Temperature
    if 'T2M' in nasa_data['parameters']:
        temp_data = nasa_data['parameters']['T2M']
        dates = list(temp_data.keys())
        temps = list(temp_data.values())
        
        axes[0].plot(dates[::3], temps[::3], 'o-', color='red', linewidth=2)
        axes[0].set_xlabel('Date')
        axes[0].set_ylabel('Temperature (°C)')
        axes[0].set_title(f'Temperature at {location_name}')
        axes[0].grid(True, alpha=0.3)
        axes[0].tick_params(axis='x', rotation=45)
    
    # Solar radiation
    if 'ALLSKY_SFC_SW_DWN' in nasa_data['parameters']:
        solar_data = nasa_data['parameters']['ALLSKY_SFC_SW_DWN']
        dates = list(solar_data.keys())
        solar = list(solar_data.values())
        
        axes[1].plot(dates[::3], solar[::3], 's-', color='orange', linewidth=2)
        axes[1].set_xlabel('Date')
        axes[1].set_ylabel('Solar Irradiance (kW/m²)')
        axes[1].set_title('Solar Radiation')
        axes[1].grid(True, alpha=0.3)
        axes[1].tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.savefig('environmental_data.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✓ Plot saved as 'environmental_data.png'")
```

## 5. Harmonic Analysis Visualization

```python
# Visualize harmonic relationships
fig, ax = plt.subplots(figsize=(14, 8))

species_colors = {
    'arabidopsis': 'green',
    'trichogramma': 'blue',
    'human': 'red'
}

y_offset = 0
for species_name, species_data in comparison['species_comparison'].items():
    for rhythm_name, period_data in species_data['periods'].items():
        period = period_data['biological_period_hours']
        harmonic = period_data['nearest_harmonic']
        is_harmonic = period_data['is_harmonic']
        
        marker = 'o' if is_harmonic else 'x'
        markersize = 12 if is_harmonic else 8
        alpha = 1.0 if is_harmonic else 0.5
        
        ax.scatter(
            period,
            y_offset,
            c=species_colors[species_name],
            marker=marker,
            s=markersize**2,
            alpha=alpha,
            label=f"{species_name}: {rhythm_name}" if y_offset < 3 else ""
        )
        
        # Add harmonic number annotation
        if is_harmonic:
            ax.annotate(
                f"n={harmonic}",
                (period, y_offset),
                xytext=(5, 5),
                textcoords='offset points',
                fontsize=8
            )
        
        y_offset += 1

ax.set_xlabel('Period (hours)', fontsize=12)
ax.set_ylabel('Biological Rhythm', fontsize=12)
ax.set_title('Harmonic Relationships with 141Hz (○ = harmonic, × = non-harmonic)', fontsize=14)
ax.set_xscale('log')
ax.grid(True, which='both', alpha=0.3)
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.savefig('harmonic_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print("✓ Plot saved as 'harmonic_analysis.png'")
```

## 6. Statistical Summary

```python
# Generate statistical summary
summary_data = []

for species_name, species_data in comparison['species_comparison'].items():
    for rhythm_name, period_data in species_data['periods'].items():
        summary_data.append({
            'Species': species_name,
            'Rhythm': rhythm_name,
            'Period (h)': period_data['biological_period_hours'],
            'Frequency (Hz)': period_data['biological_freq_hz'],
            'Harmonic': period_data['nearest_harmonic'],
            'Deviation (%)': period_data['harmonic_deviation_percent'],
            'Is Harmonic': '✓' if period_data['is_harmonic'] else '✗'
        })

df = pd.DataFrame(summary_data)
print("\nStatistical Summary of Biological Periodicities")
print("="*80)
print(df.to_string(index=False))

# Save to CSV
df.to_csv('biological_periodicities_summary.csv', index=False)
print("\n✓ Summary saved as 'biological_periodicities_summary.csv'")
```
